```python
import numpy as np
import pandas as pd
from pandera.errors import SchemaError
from pandera.pandas import Check, Column, DataFrameSchema

tickets = {
    "clarity": ["I1", "SI2", "SI1", "VS2", "VS1", "VVS2", "VVS1", "IF"],
    "color": ["D", "E", "F", "G", "H", "I", "J"],
    "cut": ["Fair", "Good", "Very Good", "Premium", "Ideal"],
}

# data directory path
diamantes_df = pd.read_parquet("data/04_feature/diamantes_clean.parquet")
diamantes_df.head()
```

```python
schema = DataFrameSchema(
    columns={
        "carat": Column(np.float32, checks=Check.gt(0)),
        "cut": Column(str, checks=Check.isin(tickets["cut"])),
        "color": Column(str, checks=Check.isin(tickets["color"])),
        "clarity": Column(str, checks=Check.isin(tickets["clarity"])),
        "depth": Column(np.float32, checks=Check.gt(0)),
        "table": Column(np.float32, checks=Check.gt(0)),
        "price": Column(np.int64, checks=Check.gt(0)),
        "x": Column(np.float32, checks=Check.gt(0)),
        "y": Column(np.float32, checks=Check.gt(0)),
        "z": Column(np.float32, checks=Check.gt(0)),
    }
)
```

```python
try:
    validated_df = schema.validate(diamantes_df, lazy=True)
    print("DataFrame validado exitosamente:")
    print(validated_df)
except SchemaError as err:
    print("Errores de validación encontrados:")
    print(err.failure_cases)
```

```python
df_diamantes_ = pd.read_csv("data/01_raw/diamantes.csv")
df_diamantes_.head()
```

```python
try:
    schema.validate(df_diamantes_, lazy=True)
except SchemaError as err:
    print("Errores de validación encontrados:")
    print(err.failure_cases)
```
