## 📚 Import  libraries

```python
# base libraries for data science
from pathlib import Path

import pandas as pd
from ydata_profiling import ProfileReport
```

## 💾 Load data

```python
DATA_DIR = Path.cwd().resolve().parents[1] / "data"
diamantes_df = pd.read_parquet(
    DATA_DIR / "02_intermediate/diamantes_type_fixed.parquet", engine="pyarrow"
)
```

## 📊 Data description

```python
diamantes_df.info()
```

```python
type_schema = {
    "cut": "categorical",
    "color": "categorical",
    "clarity": "categorical",
}
```

```python
profile = ProfileReport(
    diamantes_df,
    title="Diamantes Pandas Profiling Report",
    explorative=True,
    correlations={"auto": {"calculate": False}},
    type_schema=type_schema,
)
```

### Report embedded in the notebook

`profile.to_widgets()` errors on this ydata-profiling version (`ValueError: widget type not understood, 'overview_tabs'` — an upstream bug in the ipywidgets renderer), and a plain `HTML(profile.to_html())` / `profile.to_notebook_iframe()` gets visually mangled by VS Code/Jupyter's dark theme, since the report never declares its own `color-scheme` and ends up a mix of light and dark, some elements unreadable.

We export the report in `data/08_reporting/diamantes_pandas_profiling_report.html` to visualize it in a browser, which is the only way to get a proper rendering of the report.

###  💾 Save report to file

```python
profile.to_file(DATA_DIR / "08_reporting/diamantes_pandas_profiling_report.html")
```
